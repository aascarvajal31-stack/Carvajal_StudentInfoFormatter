```python
firstname = str(input("Enter first name: ")).strip()
lastname = str(input("Enter last name: ")).strip()
section = input("Enter your section: ")
favorite_subject = str(input("Enter your favorite subject: "))


full_name = (firstname + " " + lastname).title().strip()
name_code = full_name.upper()
section_formatted = section.upper()
subject_formatted = favorite_subject.title()
character_count = len(full_name)


student_info = [full_name, section_formatted, subject_formatted, character_count]

def student_info_display():
  for i in range(len(student_info)):
    if i == 0:
      print("NAME: ", student_info[i])
      print("NAME CODE: ", student_info[i].upper())
    elif i == 1:
      print("SECTION: ", student_info[i])
    elif i == 2:
      print("FAVORITE SUBJECT: ", student_info[i])
    elif i == 3:
      print("CHARACTER COUNT:", student_info[i])
      break

print("\n--Student Info--")
student_info_display()

```
