```python
firstname = str(input("Enter first name: "))
lastname = str(input("Enter last name: "))
section = input("Enter your section: ")
favorite_subject = str(input("Enter your favorite subject: "))

student_info = [firstname, lastname, section, favorite_subject]

def student_info_display():
  for i in range(len(student_info)):
    if i == 0:
      print("NAME: ", student_info[i], student_info[i + 1])
    elif i == 1:
      continue
    elif i == 2:
      print("SECTION: ", student_info[i])
    elif i == 3:
      print("FAVORITE SUBJECT: ", student_info[i])
      break

print("\n--Student Info--")
student_info_display()

```
